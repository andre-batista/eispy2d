```python
from eispy2d.core import (configuration as cfg, error)
import unittest
import os

class TestConfiguration(unittest.TestCase):
  def setUp(self):
    self.fileNames = ('Test1', 'Test2')
    self.config1 = cfg.Configuration(
      name=self.fileNames[0],
      wavelength=8.0,
      number_measurements=32,
      number_sources=32,
      image_size=[2.0, 2.0],
      observation_radius=5.0,
      background_permittivity=4,
      background_conductivity=0.1
    )

    self.config2 = cfg.Configuration(
        name=self.fileNames[1],
        frequency=300000000,
        number_measurements=16,
        number_sources=16,
        image_size=[2.0, 2.0],
        observation_radius=3.0,
        background_permittivity=1,
        background_conductivity=0.0,
        magnitude=1.0
    )
  def test_01_creation(self):
    self.assertIsNotNone(self.config1)
    self.assertIsNotNone(self.config2)
    print("✓ Passed creation")

  def test_02_file_saving(self):
    self.config1.save(file_path='./')
    self.config2.save(file_path='./')
    files = os.listdir('./')
    file1 = None
    file2 = None

    for f in files:
        if f == self.config1.name:
            file1 = f
        elif f == self.config2.name:
            file2 = f
    self.assertIsNotNone(file1, f"File {self.config1.name} not saved")
    self.assertIsNotNone(file2, f"File {self.config1.name} not saved")
    print("✓ Passed saving files")

  def test_03_file_loading(self):
    config1_loaded = cfg.Configuration(import_filepath='./', import_filename=self.fileNames[0])
    config2_loaded = cfg.Configuration(import_filepath='./', import_filename=self.fileNames[1])
    self.assertEqual(self.config1.name, config1_loaded.name)
    self.assertEqual(self.config1.epsilon_rb, config1_loaded.epsilon_rb)
    self.assertEqual(self.config1.lambda_b, config1_loaded.lambda_b)
    self.assertEqual(self.config1.f, config1_loaded.f)
    self.assertEqual(self.config1.sigma_b, config1_loaded.sigma_b)
    self.assertEqual(self.config1.E0, config1_loaded.E0)
    self.assertEqual(self.config1.good_conductor, config1_loaded.good_conductor)
    self.assertEqual(self.config1.perfect_dielectric, config1_loaded.perfect_dielectric)

    self.assertEqual(self.config2.name, config2_loaded.name)
    self.assertEqual(self.config2.epsilon_rb, config2_loaded.epsilon_rb)
    self.assertEqual(self.config2.lambda_b, config2_loaded.lambda_b)
    self.assertEqual(self.config2.f, config2_loaded.f)
    self.assertEqual(self.config2.sigma_b, config2_loaded.sigma_b)
    self.assertEqual(self.config2.E0, config2_loaded.E0)
    self.assertEqual(self.config2.good_conductor, config2_loaded.good_conductor)
    self.assertEqual(self.config2.perfect_dielectric, config2_loaded.perfect_dielectric)
    print("✓ Passed loading files")
  def test_04_copy(self):
    config1_copied = self.config1.copy()
    self.assertEqual(self.config1.name, config1_copied.name)
    self.assertEqual(self.config1.epsilon_rb, config1_copied.epsilon_rb)
    self.assertEqual(self.config1.lambda_b, config1_copied.lambda_b)
    self.assertEqual(self.config1.f, config1_copied.f)
    self.assertEqual(self.config1.sigma_b, config1_copied.sigma_b)
    self.assertEqual(self.config1.E0, config1_copied.E0)
    self.assertEqual(self.config1.good_conductor, config1_copied.good_conductor)
    self.assertEqual(self.config1.perfect_dielectric, config1_copied.perfect_dielectric)
    print("✓ Passed first copying test")

    config1_copied.copy(self.config2)
    self.assertEqual(self.config2.name, config1_copied.name)
    self.assertEqual(self.config2.epsilon_rb, config1_copied.epsilon_rb)
    self.assertEqual(self.config2.lambda_b, config1_copied.lambda_b)
    self.assertEqual(self.config2.f, config1_copied.f)
    self.assertEqual(self.config2.sigma_b, config1_copied.sigma_b)
    self.assertEqual(self.config2.E0, config1_copied.E0)
    self.assertEqual(self.config2.good_conductor, config1_copied.good_conductor)
    self.assertEqual(self.config2.perfect_dielectric, config1_copied.perfect_dielectric)
    print("✓ Passed second copying test")

  def test_05_compute_frequency(self):
    frequency1 = cfg.compute_frequency(wavelength=8.0) / 1000000 #MHz
    self.assertAlmostEqual(frequency1, 37.5, places=0)
    print("✓ Passed first compute frequency test")

    frequency2 = cfg.compute_frequency(wavelength=6.0,epsilon_r=1.2, mu_r=1.1, sigma=0.1) / 1000000 #MHz
    self.assertAlmostEqual(frequency2, 2.5, places=0)
    print("✓ Passed second compute frequency test")

  def test_06_compute_wavelength(self):
    self.assertAlmostEqual(cfg.compute_wavelength(frequency=37500000), 8.0, places=0)
    print("✓ Passed first compute wavelength test")

    self.assertAlmostEqual(cfg.compute_wavelength(frequency=2500000,epsilon_r=1.2, mu_r=1.1, sigma=0.1), 6.0, places=0)
    print("✓ Passed second compute wavelength test")
```

```python
from eispy2d.core import error


class TestError(unittest.TestCase):

    def test_01_missing_input_error(self):
        exc1 = error.MissingInputError('test_function', 'param1')
        self.assertIsInstance(exc1, error.Error)
        self.assertEqual(exc1.function_name, 'test_function')
        self.assertEqual(exc1.input_names, 'param1')
        self.assertIn('param1', exc1.message)
        self.assertIn('test_function', exc1.message)

        exc2 = error.MissingInputError('test_function', ['param1', 'param2'])
        self.assertEqual(exc2.function_name, 'test_function')
        self.assertEqual(exc2.input_names, ['param1', 'param2'])
        self.assertIn('param1', exc2.message)
        self.assertIn('param2', exc2.message)
        self.assertIn('test_function', exc2.message)
        print("✓ Passed MissingInputError")

    def test_02_excessive_inputs_error(self):
        exc = error.ExcessiveInputsError('test_function', ['param1', 'param2'])
        self.assertIsInstance(exc, error.Error)
        self.assertEqual(exc.function_name, 'test_function')
        self.assertEqual(exc.input_names, ['param1', 'param2'])
        self.assertIn('param1', exc.message)
        self.assertIn('param2', exc.message)
        self.assertIn('or', exc.message.lower())
        print("✓ Passed ExcessiveInputsError")

    def test_03_missing_attributes_error(self):
        exc = error.MissingAttributesError('TestClass', 'attribute_name')
        self.assertIsInstance(exc, error.Error)
        self.assertEqual(exc.class_name, 'TestClass')
        self.assertEqual(exc.attribute_name, 'attribute_name')

        print("✓ Passed MissingAttributesError")

    def test_04_wrong_type_input(self):
        exc = error.WrongTypeInput('test_function', 'param', 'int', 'str')
        self.assertIsInstance(exc, error.Error)
        self.assertEqual(exc.function_name, 'test_function')
        self.assertEqual(exc.input_name, 'param')
        self.assertEqual(exc.right_type, 'int')
        self.assertEqual(exc.wrong_type, 'str')
        self.assertIn('int', exc.message)
        self.assertIn('str', exc.message)
        self.assertIn('test_function', exc.message)
        print("✓ Passed WrongTypeInput")

    def test_05_wrong_value_input(self):
        exc = error.WrongValueInput('test_function', 'param', 'value > 0', '-1')
        self.assertIsInstance(exc, error.Error)
        self.assertEqual(exc.function_name, 'test_function')
        self.assertEqual(exc.input_name, 'param')
        self.assertEqual(exc.expect_values, 'value > 0')
        self.assertEqual(exc.given_value, '-1')
        self.assertIn('value > 0', exc.message)
        self.assertIn('-1', exc.message)
        self.assertIn('test_function', exc.message)
        print("✓ Passed WrongValueInput")

    def test_06_empty_attribute(self):
        exc = error.EmptyAttribute('TestClass', 'attribute_name')
        self.assertIsInstance(exc, error.Error)
        self.assertEqual(exc.class_name, 'TestClass')
        self.assertEqual(exc.attribute_name, 'attribute_name')
        self.assertIn('TestClass', exc.message)
        self.assertIn('attribute_name', exc.message)
        self.assertIn('empty', exc.message.lower())
        print("✓ Passed EmptyAttribute")

    def test_07_inheritance(self):
        self.assertTrue(issubclass(error.MissingInputError, error.Error))
        self.assertTrue(issubclass(error.ExcessiveInputsError, error.Error))
        self.assertTrue(issubclass(error.MissingAttributesError, error.Error))
        self.assertTrue(issubclass(error.WrongTypeInput, error.Error))
        self.assertTrue(issubclass(error.WrongValueInput, error.Error))
        self.assertTrue(issubclass(error.EmptyAttribute, error.Error))
        print("✓ Passed inheritance check")
```
